## Exercise 2
In this exercise, you’ll explore how three different machine learning models—Linear Regression, Decision Tree Regressor, and K-Nearest Neighbors (KNN) Regressor—approach the same dataset in unique ways. Don’t worry about the details of how these algorithms work just yet; we’ll dive deeper into each method in future lessons. For now, focus on understanding the input-output relationship: the models take GDP per capita as input and predict life satisfaction as output. By visualizing the fits for all three models, you’ll see how their approaches differ and gain insight into their strengths and limitations. This exercise provides a hands-on introduction to model comparison and highlights how the choice of model impacts results.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor

# Step 1: Load the dataset
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

#Preview the data
print("Dataset preview:")
print(lifesat.head())

# Step 2: Initialize three models
linear_model = LinearRegression()
tree_model = DecisionTreeRegressor(max_depth=3, random_state=42)
knn_model = KNeighborsRegressor(n_neighbors=5)

# Step 3: Train all models on the data
linear_model.fit(X, y)
tree_model.fit(X, y)
knn_model.fit(X, y)

#Print the models' scores on the training data
print("Linear Regression score:", linear_model.score(X, y))
print("Decision Tree score:", tree_model.score(X, y))
print("KNN Regressor score:", knn_model.score(X, y))

# Step 4: Generate predictions for visualization
X_plot = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
linear_predictions = linear_model.predict(X_plot)
tree_predictions = tree_model.predict(X_plot)
knn_predictions = knn_model.predict(X_plot)

# Step 5: Visualize the data and the model fits
plt.figure(figsize=(10, 6))

# Plot the original data
plt.scatter(X, y, color="black", label="Original data")

# Plot Linear Regression fit
plt.plot(X_plot, linear_predictions, color="blue", label="Linear Regression")

# Plot Decision Tree fit
plt.plot(X_plot, tree_predictions, color="green", label="Decision Tree")

# Plot KNN fit
plt.plot(X_plot, knn_predictions, color="red", label="KNN Regressor")

# Customize the plot
plt.xlabel("GDP per capita (USD)")
plt.ylabel("Life Satisfaction")
plt.title("Model Comparison: Linear Regression, Decision Tree, and KNN")
plt.legend()
plt.grid(True)
plt.show()
```

## Question 2.1 (30 pts)
Experiment with hyperparameters (e.g., max_depth for Decision Tree or n_neighbors for KNN) and observe how the fits change)

```python
# Helper library used to basically iterate through the combinations of hyperparameters chosen. This is used to simulate GridSearchCV.
import itertools

# Getting the total rows of the dataset
rows = X.shape[0]
```

```python
# For KNN, I will choose the following hyperparameters to fine tune: n neighbors, weight, and metrics. Because these are the prime hyperparameters that can substantially affect KNN's model.

# Step 1: Setting up the parameter grid (the hyperparameters to be used, and their possible values). For weights and metrics, I will be providing two different hyperparameters each, while for n_neighbors, I will be testing all of the possible values up to the total rows of the dataset.
n_neighbors = range(1,rows)
weights = ['uniform', 'distance']
metrics = ['euclidean', 'manhattan']

# Array for the results
knn_results = []

# Step 2: Iterate through the combinations of hyperparameter using itertools to find the best combination of hyperparameter that can capture the patterns in the data better.
for k, w, m in itertools.product(n_neighbors, weights, metric):
  knn = KNeighborsRegressor(n_neighbors=k, weights=w, metric=m)
  knn.fit(X, y)
  
  knn_results.append({
    'n_neighbors': k,
    'weights': w,
    'metrics': m,
    'score': knn.score(X, y),
  })

# Step 3: Converting it to a dataframe for easier interpretation and sorting of the results.
knn_df = pd.DataFrame(knn_results)
knn_df.sort_values(by='score', ascending=False, inplace=True)
knn_df
```

```python
# We can see that there are multiple combinations of n_neighbors, weights, and metric that outputs a score of perfect 1. From observation only, we can see that weights=distance produces a lot of perfect score outputs. But for the other hyperparameters, the result is a bit mixed. So we'll find the most frequent values of n_neighbors and metrics that output the perfect score of 1.

# Step 4: Find the most frequent hyperparameter used for n_neighbors and metric that produces a score output of 1. 
perfect_scores = knn_df[knn_df['score'] == 1]

print("Most frequent n_neighbor:")
print(perfect_scores['n_neighbors'].value_counts())

print("Most frequent weight:")
print(perfect_scores['weights'].value_counts())

print("\nMost frequent metric:")
print(perfect_scores['metrics'].value_counts())
```

```python
# We can see that the n_neighbors that fits the pattern of the data best is n=1, which makes sense because it essentially memorizes the dataset's points and essentially overfit the data. On the other hand, the most frequent metric is tied between eucledian and manhattan, and therefore we can observe that metric doesn't affect the model's prediction as significantly as n_neighbors and weights.

# Now, we can determine that the best combination of hyperparameter for KNN that fits the pattern best are: n_neighbors=1, weights=distance, and metric=eucledian/manhattan

# Step 5: Plotting the graph of the before and after tuning the hyperparameter of the KNN model

plt.figure(figsize=(10, 6))

# Training the new KNN Model
new_knn = KNeighborsRegressor(n_neighbors=1, weights='distance', metric='euclidean')
new_knn.fit(X, y)

print("Old KNN Regressor score:", knn_model.score(X, y))
print("New KNN Regressor score:", new_knn.score(X, y))

# Generating the prediction for visualization
new_knn_predictions = new_knn.predict(X_plot)

# Plot the original data
plt.scatter(X, y, color="black", label="Original data")

# Plot both old and new KNN fit
plt.plot(X_plot, knn_predictions, color="red", label="Old KNN Regressor")
plt.plot(X_plot, new_knn_predictions, color="blue", label="New KNN Regressor")

# Additional customization
plt.xlabel("GDP per capita (USD)")
plt.ylabel("Life Satisfaction")
plt.title("Model Comparison: old and new KNN models")
plt.legend()
plt.grid(True)
plt.show()
```
